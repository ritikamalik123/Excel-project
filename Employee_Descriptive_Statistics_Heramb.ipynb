# Descriptive Statistics Case Study – Employee Dataset

This notebook studies the employee dataset using **Pandas, NumPy, Matplotlib and Seaborn**.

The analysis covers:
1. Loading and understanding the data
2. Visualizing important features
3. Filtering employees using given conditions


## Task 1 – Load and Study the Data

```python
# Import the required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Read the employee dataset

df = pd.read_csv("Employee Dataset.csv")

df.head()
```

```python
# Check the dimensions of the dataset

df.shape
```

```python
# Check the index of the dataframe

df.index
```

```python
# Display the column names

df.columns
```

```python
# Get basic information about the dataset

df.info()
```

```python
# Check for missing values

df.isnull().sum()
```

```python
# Get descriptive statistics for numerical columns

df.describe()
```

### Observation

The dataset contains **50 employees and 6 columns**. There are no missing values in any of the columns.


## Task 2 – Visualize the Distributions

```python
# Age versus salary

plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x="age", y="salary", alpha=0.7)

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

**Observation:** Salary varies across different age groups, and the plot does not show one clear linear relationship between age and salary.


```python
# Healthy eating versus active lifestyle

plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x="healthy_eating",
    y="active_lifestyle",
    alpha=0.7
)

plt.title("Healthy Eating vs Active Lifestyle")
plt.xlabel("Healthy Eating Score")
plt.ylabel("Active Lifestyle Score")
plt.show()
```

**Observation:** The employees have different combinations of healthy eating and active lifestyle scores, with the points spread across the plot.


```python
# Healthy eating versus salary

plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x="healthy_eating", y="salary", alpha=0.7)

plt.title("Healthy Eating vs Salary")
plt.xlabel("Healthy Eating Score")
plt.ylabel("Salary")
plt.show()
```

**Observation:** Employees with similar healthy eating scores can have different salaries, so the two variables do not show a simple direct relationship.


```python
# Count of employees in each group

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="groups", hue="groups", legend=False)

plt.title("Number of Employees in Each Group")
plt.xlabel("Group")
plt.ylabel("Number of Employees")
plt.show()

df["groups"].value_counts().sort_index()
```

**Observation:** Group A has the largest number of employees, followed by group O.


```python
# Distribution of employee salaries

plt.figure(figsize=(9, 5))
sns.histplot(data=df, x="salary", bins=10, kde=True)

plt.title("Distribution of Employee Salaries")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
```

**Observation:** Most salary values are concentrated in the lower and middle part of the range, while a few employees have relatively high salaries.


## Task 3 – Subset the Data Based on Thresholds

```python
# Employees with healthy eating score greater than 8

healthy_employees = df[df["healthy_eating"] > 8]
healthy_employees
```

```python
# Employees with salary less than 1000

low_salary = df[df["salary"] < 1000]
low_salary
```

```python
# Employees satisfying both conditions

result = df[
    (df["healthy_eating"] > 8) &
    (df["salary"] < 1000)
]

result
```

### Final Observation

Only one employee satisfies both conditions: **healthy eating score greater than 8** and **salary below 1000**.

This employee belongs to group A, is 62 years old, has a healthy eating score of 9, an active lifestyle score of 6, and a salary of 700.
