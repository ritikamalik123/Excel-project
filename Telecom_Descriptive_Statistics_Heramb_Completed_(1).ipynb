# Descriptive Statistics Case Study – Telecom Dataset

This notebook studies the provided **Telecom Dataset** using **Pandas, NumPy, Matplotlib and Seaborn**.

The analysis covers:
1. Loading and understanding the data
2. Checking dimensions, columns, data types and missing values
3. Computing descriptive statistics
4. Visualizing important numerical and categorical features
5. Filtering products using practical conditions
6. Summarizing the main findings


## Task 1 – Load and Study the Data

```python
# Import the required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
# Read the telecom dataset

df = pd.read_csv("Telecom Dataset.csv")

df.head()
```

```python
# Check the dimensions of the dataset
df.shape
```

```python
# Check the index of the dataframe
df.index
```

```python
# Display the column names
df.columns
```

```python
# Get basic information about the dataset
df.info()
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Check for duplicate rows
df.duplicated().sum()
```

```python
# Get descriptive statistics for numerical columns
df.describe()
```

### Observation

The dataset contains **50 products and 17 columns**. The dataset has no missing values and no duplicate rows. It contains four binary feature columns (`Blue`, `Wi_Fi`, `Tch_Scr`, and `Ext_Mem`) together with numerical specifications such as processor-related values, memory, battery power, RAM, depth, weight and price.


## Task 2 – Explore Categorical Features

```python
# Count products by bluetooth

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Blue", hue="Blue", legend=False)

plt.title("Bluetooth Availability")
plt.xlabel("Bluetooth")
plt.ylabel("Number of Products")
plt.show()

df["Blue"].value_counts()
```

```python
# Count products by wi-fi

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Wi_Fi", hue="Wi_Fi", legend=False)

plt.title("Wi-Fi Availability")
plt.xlabel("Wi-Fi")
plt.ylabel("Number of Products")
plt.show()

df["Wi_Fi"].value_counts()
```

```python
# Count products by touch screen

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Tch_Scr", hue="Tch_Scr", legend=False)

plt.title("Touch Screen Availability")
plt.xlabel("Touch Screen")
plt.ylabel("Number of Products")
plt.show()

df["Tch_Scr"].value_counts()
```

```python
# Count products by external memory

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="Ext_Mem", hue="Ext_Mem", legend=False)

plt.title("External Memory Availability")
plt.xlabel("External Memory")
plt.ylabel("Number of Products")
plt.show()

df["Ext_Mem"].value_counts()
```

### Observation

Among the 50 products, **27 have Bluetooth**, **42 have Wi-Fi**, **45 have a touch screen**, and **38 support external memory**. These counts show that Wi-Fi and touch-screen support are common in this sample.


## Task 3 – Visualize Numerical Relationships

```python
# RAM vs Price

plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x="RAM", y="Price", alpha=0.75)

plt.title("RAM vs Price")
plt.xlabel("RAM")
plt.ylabel("Price")
plt.show()
```

```python
# Internal Memory vs Price

plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x="Int_Mem", y="Price", alpha=0.75)

plt.title("Internal Memory vs Price")
plt.xlabel("Internal Memory")
plt.ylabel("Price")
plt.show()
```

```python
# Battery Power vs Price

plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x="Bty_Pwr", y="Price", alpha=0.75)

plt.title("Battery Power vs Price")
plt.xlabel("Battery Power")
plt.ylabel("Price")
plt.show()
```

```python
# Weight vs Price

plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x="Weight", y="Price", alpha=0.75)

plt.title("Weight vs Price")
plt.xlabel("Weight")
plt.ylabel("Price")
plt.show()
```

### Observation

The scatter plots show how price varies across RAM, internal memory, battery power and weight. The points are spread across a wide range of prices, so individual hardware specifications should be considered together when examining product price.


## Task 4 – Distribution of Product Prices

```python
# Distribution of product prices

plt.figure(figsize=(9, 5))
sns.histplot(data=df, x="Price", bins=10, kde=True)

plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Number of Products")
plt.show()
```

```python
# Box plot for price

plt.figure(figsize=(9, 3))
sns.boxplot(x=df["Price"])

plt.title("Box Plot of Product Prices")
plt.xlabel("Price")
plt.show()
```

### Observation

The product prices range from **3297 to 149000**. The mean price is higher than the median because the dataset contains some substantially more expensive products.


## Task 5 – Descriptive Statistics of Key Variables

```python
# Mean, median, minimum and maximum for selected numerical variables

summary = df[["RAM", "Int_Mem", "Bty_Pwr", "Weight", "Price"]].agg(
    ["mean", "median", "min", "max"]
)

summary
```

```python
# Correlation matrix for numerical variables

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=True, fmt=".2f", cmap="coolwarm")

plt.title("Correlation Matrix of Numerical Features")
plt.show()
```

### Observation

The summary table provides central tendency and range information for the main numerical variables. The correlation matrix helps identify linear relationships among numerical features, while correlation should not be interpreted as proof of causation.


## Task 6 – Subset the Data Based on Conditions

```python
# Products with RAM greater than 8

high_ram = df[df["RAM"] > 8]
high_ram
```

```python
# Products with price less than 10000

low_price = df[df["Price"] < 10000]
low_price
```

```python
# Products with both RAM greater than 8 and price below 10000

result = df[
    (df["RAM"] > 8) &
    (df["Price"] < 10000)
]

result
```

```python
# Products that have Wi-Fi, touch screen and external memory

feature_rich = df[
    (df["Wi_Fi"] == "yes") &
    (df["Tch_Scr"] == "yes") &
    (df["Ext_Mem"] == "yes")
]

feature_rich
```

### Observation

The filtering operations demonstrate how Pandas can be used to identify products that satisfy one or more numerical or categorical conditions. The resulting tables show the complete product records that meet each condition.
